# Journal 1 — Reflecting on My First Numerical Solve

Transitioning from paper-and-pencil physics to writing a numerical simulation brought a few surprising insights:

1. **The Choppiness of Data:**
I was surprised by how much the choice of time step ($dt$) matters. On paper, time is continuous, and equations like $y = h - \frac{1}{2}gt^2$ are always perfectly exact. In the simulation, turning $dt$ up to $0.5\text{s}$ completely broke the physics and overshot the range by $14\text{ meters}$. I wouldn't have thought it would matter quite that much.

2. **The Impact of Drag:**
After several semester basically actinglike drag isn't a thing, or that it's negligible, I finally get to see how drastic of a change it can really make. Seeing how air resistance completely morphs a perfect symmetric parabola into an asymmetric shape with a steep cliff at the end was fascinating. It surprised me that a simple fastball loses almost $40\%$ of its distance just from hitting air molecules, and that the best angle drops below $45^\circ$ to $40^\circ$. You can't calculate that easily (if at all) by hand, which makes me see why coding is such a vital tool for real-world problems.

3. **Experience:**
After getting through the setup and being loaned a pretty good computer, I wasn't as miseraable with coding as I thought I would be. There were even some aspects that were kind of fun.
